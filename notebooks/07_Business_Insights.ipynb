```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv(r"../data/processed/feature_engineered_data.csv")
df.head()
```

```python
df["Promo"].value_counts()
```

## KPIs

```python
kpis = {
    "Total Sales": df["Sales"].sum(),
    "Average Daily Sales": df["Sales"].mean(),
    "Number of Stores": df["Store"].nunique(),
    "Average Promotion Sales": df[df["Promo"]==1]["Sales"].mean(),
    "Average Non-Promotion Sales": df[df["Promo"]==0]["Sales"].mean()
}

pd.DataFrame(kpis.items(), columns=["KPI","Value"])
```

# Inventory Planning
| Forecast  | Recommendation          |
| --------- | ----------------------- |
| Very High | Increase inventory      |
| High      | Moderate stock increase |
| Medium    | Maintain normal stock   |
| Low       | Reduce inventory        |


## Promotion Strategy
Promotions consistently increase predicted sales. Stores with historically strong promotion response should be prioritized for future campaigns.

## Store-Level Recommendations

```python
store_sales = (
    df.groupby("Store")["Sales"]
    .mean()
    .sort_values(ascending=False)
    .reset_index()
)
store_sales
```

 `High Performers`
- Maintain stock levels.
- Expand successful promotions.

`Average Performers`
- Optimize product assortment.
- Evaluate local competition.

`Low Performers`
- Review pricing.
- Assess store operations.
- Investigate demand patterns.

## Seasonal Planning

- **Peak Months**: January and March 
- **Lowest Months**: August, September, and October
- **Summer Slump**: Clear decline during summer months (July-October)
- **Recommendation**: Plan additional marketing and promotions during August-September to combat seasonal dip

## Staffing Recommendations
- High Forecast ➡️  More Cashiers ➡️ More Inventory Staff ➡️ Longer Store Hours
- Reduce Staffing ➡️ Reduce Operational Cost

## Risk Analysis

`Limitations`
- Forecasts rely on historical sales patterns.
- Unexpected events (pandemics, policy changes, extreme weather) are not modeled.
- Inventory shortages are not explicitly represented.
- Economic indicators and inflation were not included.
