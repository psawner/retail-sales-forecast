## Business Problem

Rossmann operates thousands of retail stores.

Management wants to accurately forecast daily sales so they can:

- optimize inventory
- schedule staff
- plan promotions
- reduce stockouts
- reduce waste

## Objectives

- Forecast future sales.
- Identify drivers of sales.
- Understand promotion effectiveness.
- Analyze seasonality.
- Recommend inventory planning.

## KPIs

Instead of just RMSE, we'll also discuss business KPIs like:

- Daily Sales
- Weekly Sales
- Revenue Growth
- Promotion Lift
- Sales per Store
- Average Daily Sales
- Forecast Accuracy
- Inventory Coverage (estimated)
